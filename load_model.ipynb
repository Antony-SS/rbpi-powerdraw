```python
import os
import shutil
from ultralytics import YOLO

def export_and_rename(model_name, output_name, img_size):
    print(f"\n--- Exporting {output_name} ({img_size}x{img_size}) ---")
    
    # 1. Load Model
    model = YOLO(model_name)
    
    # 2. Export (Always creates 'yolov8n_ncnn_model')
    # We use half=True for that 2x speedup on Pi
    model.export(format='ncnn', half=True, imgsz=img_size)
    
    # 3. Rename output
    default_name = f"{model_name.replace('.pt', '')}_ncnn_model"
    
    if os.path.exists(output_name):
        shutil.rmtree(output_name) # Clean up old export if exists
        
    os.rename(default_name, output_name)
    print(f"✅ Success! Renamed to: {output_name}")

# --- RUN THE EXPORTS ---
# Run this on your laptop!

# 1. Export the Fast Version
export_and_rename('yolov8n.pt', 'yolo_320', 320)

# 2. Export the Accurate Version
export_and_rename('yolov8n.pt', 'yolo_640', 640)

print("\nDONE. Copy 'yolo_320' and 'yolo_640' folders to your Pi.")
```
